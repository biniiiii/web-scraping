```python
import requests
import pandas as pd

url = "https://gw.jeevee.com/v1/categories/33/products?page=1&item_per_page=24"
headers = {"User-Agent": "MyProject/1.0 (binii@example.com)"}

r = requests.get(url, headers=headers)
# print(r.json())
```

```python
import pandas as pd

data = r.json()
products = data['data']  # the actual product list

records = []
for p in products:
    records.append({
        'name': p['label'],
        'brand': p['brand']['name'],
        'price': p['price'],
        'discount': p['discount'],
        'rating': p['review_and_rating']['avg_rating'],
        'rating_count': p['review_and_rating']['rating_count'],
        'sold_out': p['sold_out'],
    })

df = pd.DataFrame(records)
print(df.head())
```

```python
all_products = []

for page in range(1, 24):  # total_pages = 23
    url = f"https://gw.jeevee.com/v1/categories/33/products?page={page}&item_per_page=24"
    r = requests.get(url, headers=headers)
    data = r.json()

    if 'data' not in data:
        print(f"Skipping page {page} due to API error: {data.get('error', 'Unknown error')}")
        continue

    for p in data['data']:
        all_products.append({
            'name': p['label'],
            'brand': p['brand']['name'],
            'price': p['price'],
            'discount': p.get('discount', 0), # Use .get() to provide a default value if 'discount' key is missing
            'rating': p['review_and_rating']['avg_rating'],
            'rating_count': p['review_and_rating']['rating_count'],
            'sold_out': p['sold_out'],
        })

    print(f"Page {page} done")

df = pd.DataFrame(all_products)
df.to_csv("jeevee_serums.csv", index=False)
print(f"Total: {len(df)} products saved!")
```
